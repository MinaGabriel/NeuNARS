# MuSiQue · NARS + DeepSeek-V3 (local)

DeepSeek-V3 (the teacher) writes the triples, aliases, plan and rules; ONA derives the answer. Same facts and
k = 0, 4, 8, 16 as every other system. Needs the ONA binary (`ONA_PATH`, commit `dc4efd0`).

Answers and full NARS traces go to `results/`; running again only runs what is missing. Slow (about 1–2 minutes
per answer, run in parallel). Cost for 200 questions: about $10–15 for MuSiQue, less for the others.

```python
import sys
sys.path.insert(0, "../common")
import experiment as ex

DATASET = "MuSiQue"
N_QUESTIONS = 200             # the first N questions; 5 for a quick check
K_VALUES = ex.K_VALUES[DATASET]   # 0, 4, 8, 16
WORKERS = 12
ONA_PATH = "/Users/mina/Developer/OpenNARS-for-Applications/NAR"

questions, question_hash = ex.load_questions("questions.pkl", N_QUESTIONS)
client = ex.deepinfra()
ex.check_models([(client, "deepseek-ai/DeepSeek-V3"), (client, ex.JUDGE_MODEL)])
systems = {"NARS + DeepSeek-V3": (client, "deepseek-ai/DeepSeek-V3", ONA_PATH)}
```

```python
ex.run_all(questions, question_hash, "results", systems, K_VALUES, workers=WORKERS)
```
